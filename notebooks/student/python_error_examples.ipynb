# Python のエラー例と直し方

本編・付録の学生用 Notebook でよく見かける Python のエラーを、**見出し → エラーが出るコードと出力 → 意味と直し方 → 修正済みコードと出力**の順で示します。各組は独立しています。実行する場合は、見たいセルを選んで Shift + Enter を押してください。

エラーメッセージは、まず**最後の行**でエラー名と説明を読みます。次に矢印（`---->`）や印（`^`）を見て、示された行とその直前を確かめます。

## 例1　変数名のタイプミス

```python
# 例1（エラー）｜変数名のタイプミス
import math

error_ratio = 4.0
p_pair = math.log2(error_rato)
print(f"収束次数 = {p_pair:.1f}")
```

**意味**　`NameError: name 'error_rato' is not defined` は、Python が `error_rato` という名前を見つけられなかったという意味です。

**直し方**　値を代入した行の名前は `error_ratio` です。`math.log2(...)` に渡す名前を同じ綴りに直します。綴りが合っている場合は、その名前を定義したセルを実行済みか確認します。

```python
# 例1（修正後）
import math

error_ratio = 4.0
p_pair = math.log2(error_ratio)
print(f"収束次数 = {p_pair:.1f}")
```

## 例2　字下げが深すぎる

```python
# 例2（エラー）｜字下げが深すぎる
def trapezoid_area(left, right, width):
    mean_height = (left + right) / 2
        return mean_height * width

print(trapezoid_area(2.0, 4.0, 1.0))
```

**意味**　`IndentationError: unexpected indent` は、その位置で予期しない字下げがあるという意味です。エラー例では `return` の行が、その上の行より4スペース深くなっています。

**直し方**　同じ関数の中で続けて実行する2行を、同じ深さにそろえます。`return` の前の空白を半角スペース4個にします。`expected an indented block` と出たときは、`def` や `for` の次の行に必要な字下げを補います。

```python
# 例2（修正後）
def trapezoid_area(left, right, width):
    mean_height = (left + right) / 2
    return mean_height * width

print(trapezoid_area(2.0, 4.0, 1.0))
```

## 例3　タブとスペースの混在

```python
# 例3（エラー）｜タブとスペースの混在
def trapezoid_area_with_tab(left, right, width):
    mean_height = (left + right) / 2
	return mean_height * width

print(trapezoid_area_with_tab(1.0, 3.0, 2.0))
```

**意味**　`TabError: inconsistent use of tabs and spaces in indentation` は、行頭でタブとスペースが混在しているという意味です。エラー例では `mean_height` の行はスペース4個、`return` の行はタブ1個です。画面では同じ深さに見えることがあります。

**直し方**　`return` の手前の空白を削除し、半角スペース4個を入れます。

```python
# 例3（修正後）
def trapezoid_area_with_tab(left, right, width):
    mean_height = (left + right) / 2
    return mean_height * width

print(trapezoid_area_with_tab(1.0, 3.0, 2.0))
```

## 例4　準備セルの import を実行していない

```python
# 例4（エラー）｜準備セルの import を実行していない
x = np.array([0, 1, 2])
plt.plot(x, x**2, "o-")
plt.show()
```

**意味**　`NameError: name 'np' is not defined` は、今のカーネルで `np` という名前がまだ使える状態になっていないという意味です。`np` は NumPy、`plt` は Matplotlib の `pyplot` に付けた別名です。NumPy だけを読み込んだ状態では、グラフを描く行で `NameError: name 'plt' is not defined` が出ます。

**直し方**　授業用 Notebook では、冒頭の `import numpy as np` と `import matplotlib.pyplot as plt` がある準備セルを実行してから、エラーが出たセルを再実行します。カーネルを再起動したあとも、準備セルを実行し直します。下の修正済みセルは必要な `import` を含めているので、単独で実行できます。

`import` の行で `ModuleNotFoundError: No module named 'numpy'` などが出た場合は、現在の実行環境からライブラリが見つかっていません。授業用の Python カーネルを選んでいるか確認します。

エラーを再現するときは、カーネルを再起動してから直前のエラーセルを実行してください。

```python
# 例4（修正後）
import numpy as np
import matplotlib.pyplot as plt

x = np.array([0, 1, 2])
plt.plot(x, x**2, "o-")
plt.show()
```

## 例5　余分な閉じ括弧

```python
# 例5（エラー）｜余分な閉じ括弧
total = sum([1, 2])
print(total))
```

**意味**　`SyntaxError: unmatched ')'` は、対応する開き括弧がない閉じ括弧があるという意味です。`^` が余分な括弧の位置を示します。

**直し方**　`print(total))` には開き括弧が1個、閉じ括弧が2個あります。末尾の `)` を1つ削除します。

```python
# 例5（修正後）
total = sum([1, 2])
print(total)
```

## 例6　全角の括弧

```python
# 例6（エラー）｜全角の括弧
values = [1, 2]
total = sum（values)
print(total)
```

**意味**　`SyntaxError: invalid character '（' (U+FF08)` は、全角の開き括弧がこの位置で使えないことを示します。`^` とメッセージに出ている文字を確認します。

**直し方**　`sum（values)` の全角の `（` を半角の `(` に置き換えます。

```python
# 例6（修正後）
values = [1, 2]
total = sum(values)
print(total)
```

## 例7　代入式の末尾のカンマ

```python
# 例7（エラー）｜代入式の末尾のカンマ
p_pair = 2.0,
print(f"収束次数 = {p_pair:.1f}")
```

**意味**　`TypeError: unsupported format string passed to tuple.__format__` は、タプルに対して使えない表示形式を指定したという意味です。`p_pair = 2.0,` と書くと、`p_pair` には数値の代わりに1要素のタプル `(2.0,)` が入ります。エラーが出るのは、次の行で `:.1f` を使って小数第1位まで表示しようとしたときです。

**直し方**　表示の行から値の出どころをたどり、代入した行の末尾のカンマを削除します。

```python
# 例7（修正後）
p_pair = 2.0
print(f"収束次数 = {p_pair:.1f}")
```

## 例8　学生版の入力欄が未入力

```python
# 例8（エラー）｜学生版の入力欄が未入力
def trapezoid_with_blank(left, right, width):
    area = width * (left + right) / 2
    raise NotImplementedError("例8の return 行を入力してください。")

print(trapezoid_with_blank(2.0, 4.0, 2.0))
```

**意味**　`NotImplementedError` は、このセルでは入力欄が残っていることを知らせるために出しています。エラー出力には、関数を呼んだ行と、関数内で実行が止まった `raise` の行が表示されます。

**直し方**　行頭の字下げを残し、`raise` の行を `return area` に置き換えます。定義と呼び出しが別のセルにある場合は、まず定義したセルを実行し、そのあとで呼び出すセルを実行します。

```python
# 例8（修正後）
def trapezoid_with_blank(left, right, width):
    area = width * (left + right) / 2
    return area

print(trapezoid_with_blank(2.0, 4.0, 2.0))
```

## 例9　関数の return の書き忘れ

```python
# 例9（エラー）｜関数の return の書き忘れ
def trapezoid_area_from_endpoints(left, right, width):
    area = width * (left + right) / 2

area_num = trapezoid_area_from_endpoints(2.0, 4.0, 1.0)
area_exact = 3.0
print("絶対誤差:", abs(area_num - area_exact))
```

**意味**　`TypeError: unsupported operand type(s) for -: 'NoneType' and 'float'` は、引き算の一方が `None`、もう一方が小数型（`float`）であることを示します。この例では関数内で面積を計算していますが、値を返さずに関数が終わるため、`area_num` に `None` が入ります。エラーが出るのは、後で解析値との差を取る行です。

**直し方**　`area_num` の値を作った関数までさかのぼり、計算した面積を `return area` で返します。第1章の入力 1-01 でも、積分値を `return` で返します。定義と呼び出しが別のセルにある場合は、修正した関数の定義セルを先に実行します。

```python
# 例9（修正後）
def trapezoid_area_from_endpoints(left, right, width):
    area = width * (left + right) / 2
    return area

area_num = trapezoid_area_from_endpoints(2.0, 4.0, 1.0)
area_exact = 3.0
print("絶対誤差:", abs(area_num - area_exact))
```

## 例10　配列の形が合わない

```python
# 例10（エラー）｜配列の形が合わない
import numpy as np

# 各行は、左から x, y 座標。粗い軌道の出力時刻は 0, 1, 2。
x_coarse = np.array([[0.0, 0.0], [1.0, 1.0], [2.0, 2.0]])
# 細かい軌道の出力時刻は 0, 0.5, 1, 1.5, 2。
x_fine = np.array([[0.0, 0.0], [0.5, 0.5], [1.0, 1.125], [1.5, 1.5], [2.0, 2.25]])

dx = x_coarse - x_fine
print("位置の差の配列の形:", dx.shape)
print(dx)
```

**意味**　`ValueError: operands could not be broadcast together with shapes (3,2) (5,2)` は、この2つの配列をそのまま要素ごとに引けないことを示します。`x_coarse.shape` は `(3, 2)`、`x_fine.shape` は `(5, 2)` です。この例では各行が1つの時刻、2列がその時刻の座標を表します。

**直し方**　付録Cの入力 C-02 と同じように、物理的に同じ時刻の位置を比べます。細かい軌道の時間刻みは粗い軌道の半分で、開始時刻は同じです。`x_fine[::2]` で先頭から1行おきに取り出すと、時刻 0、1、2 の座標がそろいます。この3行を `x_coarse` から引きます。

```python
# 例10（修正後）
import numpy as np

# 各行は、左から x, y 座標。粗い軌道の出力時刻は 0, 1, 2。
x_coarse = np.array([[0.0, 0.0], [1.0, 1.0], [2.0, 2.0]])
# 細かい軌道の出力時刻は 0, 0.5, 1, 1.5, 2。
x_fine = np.array([[0.0, 0.0], [0.5, 0.5], [1.0, 1.125], [1.5, 1.5], [2.0, 2.25]])

dx = x_coarse - x_fine[::2]
print("位置の差の配列の形:", dx.shape)
print(dx)
```
